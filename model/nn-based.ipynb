```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from sklearn.metrics import classification_report
```

```python
input_dim = 1000
output_dim = 500
max_size = 200
```

```python
#rnn based model

rnn_model = keras.Sequential()
rnn_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
rnn_model.add(keras.layers.SimpleRNN(110))
rnn_model.add(keras.layers.Dropout(0.8))
rnn_model.add(keras.layers.Dense(1, activation='sigmoid'))
rnn_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#LSTM based model

lstm_model = keras.Sequential()
lstm_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
lstm_model.add(keras.layers.LSTM(128,return_sequences=True))
lstm_model.add(keras.layers.Dropout(0.8))
lstm_model.add(keras.layers.LSTM(64,return_sequences=True))
lstm_model.add(keras.layers.Dropout(0.5))
lstm_model.add(keras.layers.LSTM(32,return_sequences=True))
lstm_model.add(keras.layers.Dropout(0.5))
lstm_model.add(keras.layers.LSTM(16,return_sequences=True))
lstm_model.add(keras.layers.Dropout(0.5))
lstm_model.add(keras.layers.LSTM(8))
lstm_model.add(keras.layers.Dense(1, activation='sigmoid'))
lstm_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#GRU Based Model

gru_model = keras.Sequential()
gru_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
gru_model.add(keras.layers.Bidirectional(keras.layers.GRU(100)))
gru_model.add(keras.layers.BatchNormalization())
gru_model.add(keras.layers.Dense(1, activation='sigmoid'))
gru_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

```

```python
models = {
    'RNN': rnn_model,
    'LSTM': lstm_model,
    'GRU': gru_model
}
```

```python
X_train = np.load('data/X_train_parsbert1.npy')
X_test = np.load('data/X_test_parsbert1.npy')
y_train = np.load('data/y_train.npy')
y_test = np.load('data/y_test.npy')
```

```python
histories = []

for k, v in models.items():


  print(k)
  print("="*70)
  hist = v.fit(X_train, y_train, epochs=30, batch_size=32, validation_data=(X_test, y_test))

  histories.append(hist)


```
