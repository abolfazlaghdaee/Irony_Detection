```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
```

```python
input_dim = 1000
output_dim = 100
max_size = 100
```

```python
#rnn based model

rnn_model = keras.Sequential()
rnn_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
rnn_model.add(keras.layers.SimpleRNN(64))
rnn_model.add(keras.layers.Dense(1, activation='sigmoid'))
rnn_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#LSTM based model

lstm_model = keras.Sequential()
lstm_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
lstm_model.add(keras.layers.LSTM(64))
lstm_model.add(keras.layers.Dense(1, activation='sigmoid'))
lstm_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#GRU Based Model

gru_model = keras.Sequential()
gru_model.add(keras.layers.Embedding(input_dim, output_dim, input_length= max_size))
gru_model.add(keras.layers.GRU(64))
gru_model.add(keras.layers.Dense(1, activation='sigmoid'))
gru_model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
models = {
    'RNN': rnn_model,
    'LSTM': lstm_model,
    'GRU': gru_model
}
```
